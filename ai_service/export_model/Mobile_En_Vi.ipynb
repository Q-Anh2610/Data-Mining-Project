```python
!pip install -q onnxruntime

from onnxruntime.quantization import quantize_dynamic, QuantType
import os
import shutil

output_dir = "onnx_model_quantized"
os.makedirs(output_dir, exist_ok=True)

print("1. Đang ép cân phần Não Nghe (Encoder) từ 32-bit xuống 8-bit...")
quantize_dynamic(
    model_input="onnx_model/encoder_model.onnx",
    model_output=f"{output_dir}/encoder_model.onnx",
    weight_type=QuantType.QUInt8
)

print("2. Đang ép cân phần Không Gian Nhớ (Decoder) từ 32-bit xuống 8-bit...")
quantize_dynamic(
    model_input="onnx_model/decoder_model.onnx",
    model_output=f"{output_dir}/decoder_model.onnx",
    weight_type=QuantType.QUInt8
)

print("3. Đang dọn dẹp và gom các file từ điển (Tokenizer) sang nhà mới...")
for file in os.listdir("onnx_model"):
    if not file.endswith(".onnx"):
        shutil.copy(os.path.join("onnx_model", file), os.path.join(output_dir, file))

print("\nHOÀN THÀNH ÉP CÂN! Dưới đây là hành trang cho Dev Mobile:")
total_size = 0
for file in os.listdir(output_dir):
    size_mb = os.path.getsize(os.path.join(output_dir, file)) / (1024 * 1024)
    total_size += size_mb
    print(f"  - {file} ({size_mb:.2f} MB)")

print(f"\nTỔNG DUNG LƯỢNG MỚI: {total_size:.2f} MB")
```
